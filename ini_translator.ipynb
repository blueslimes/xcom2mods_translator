# ini 檔案翻譯器

將 ini 檔中的 value 的字串內容翻譯成其他語言，並保持原本的設定格式不變。

```ini
; ini檔格式
[Section]
key = value
```


```ini
[example]
language = "English"
; 會被翻譯成

[example]
language = "英文"
```



````python
from IPython.display import clear_output
import re

import googletrans
translator = googletrans.Translator()
````

````python
#選擇要翻譯的檔案
path = '2357292260/Localization/XComGame'

# 開啟檔案並建立一個副檔名為 cht 的新檔案，將翻譯後的內容寫入新檔案中
# 檔案副檔名中 int 代表英文、 cht 代表繁體中文、 chs 代表簡體中文
# encoding 為編碼格式，可以使用各種文字編輯器查看(推薦vscode)，通常為 utf-16-le、utf-8
with open('../'+path+'.int', 'r',encoding="utf-16-le") as f0 ,open('未檢查/'+path+'.cht', 'w+',encoding="utf-16-le") as f1 :
    
    file = f0.read()
    lines = file.split("\n")
    for line in lines:
        sentences= line.split("\"")
        try:
            sentence=sentences[1]
            #print(sentence)

            # split 保留 < 和 > 作為獨立項目，方便跳過標籤內容。
            text = re.split(r"([<>])",sentence)
            text_len=len(text)
            #print(text)

             # 每組文字／標籤以四個項目排列；只翻譯文字項目，避免破壞遊戲標籤。
            for s in range(0,text_len,4):
                print(text[s])
                text[s] = translator.translate(text[s], dest='zh-tw').text # 翻譯成繁體中文
                print(text[s])


            textmix =""    
            for i in text:
                textmix += i

            sentences[1] = '\"' + textmix + '\"'
            #clear_output(wait=True)


        except :
            # 無法解析或翻譯的行維持原內容，避免中斷整個檔案的轉換。
            pass 
        finally:
             # 輸出原行的各部分，讓未處理的設定與註解也保留在結果中。
            for sentence in sentences:
                f1.write(sentence)
            f1.write("\n")

````
